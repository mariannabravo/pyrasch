```python
import irw
import numpy as np
import pandas as pd
```

```python
# downloading the appropriate dataset, located through filtering directly on the irw website
gilbert = irw.fetch("gilbert_meta_2")
```

```python
# the dataset is downloaded as a pandas DataFrame, according to the python manual 
type(gilbert)
```

```python
# getting a better idea of what the DataFrame looks like, especially column labels
print(gilbert.shape)
gilbert.head(10)
```

```python
# relevant metadata for rasch analysis

no_items, no_students = gilbert["item"].nunique(), gilbert["id"].nunique()
no_missing_values = gilbert["resp"].isna().sum()
item_p = gilbert.groupby("item")["resp"].mean()
print(f"Number of items: {no_items}")
print(f"Number of students: {no_students}")
print(f"Are there missing responses?: {"Yes" if no_missing_values > 0 else "No"}")

```

```python
# obtaining item text through an irw command 
gilbert_items = irw.itemtext("gilbert_meta_2")
# displaying the table to get an idea of how the data has been laid out
gilbert_items.head()
```

```python
# isolating the columns with item text and relevant ids

gilbert_items = irw.itemtext("gilbert_meta_2")
item_questions = (
    gilbert_items[["section_id", "item", "item_text"]]
    .drop_duplicates()
    .sort_values("item")
)

# printing the item text clearly

for _, row in item_questions.iterrows():
    print(f"\nItem {row["item"]} — {row["section_id"]}")
    print(row["item_text"])
```

```python
# plotting a distribution of person scores

import matplotlib.pyplot as plt

student_scores = gilbert.groupby("id")["resp"].sum()

plt.hist(student_scores, bins = np.arange(-0.5, 21.5, 1), edgecolor = "black")

plt.xlabel("Total score")
plt.ylabel("Number of students with score")
plt.title("Distribution of Total Scores")

plt.show()
```

```python
item_scores = gilbert.groupby("item")["resp"].mean()

plt.bar(item_scores.index.astype(str), item_scores.values, color = "green")

plt.xlabel("Item")
plt.ylabel("Proportion correct")
plt.title("Observed Proportion Correct by Item")
plt.show()
```

```python
# making a response dataframe and printing it to ensure it is in the correct form to later convert to a numpy array
response_dataframe = gilbert.pivot(index = "id", columns = "item", values = "resp")
response_dataframe.head()
```
